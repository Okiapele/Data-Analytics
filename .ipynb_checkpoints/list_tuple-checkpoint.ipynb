In Python, *both `list` and `tuple` are used to store collections of items*, but they behave very differently.

Here’s the breakdown:

### *1. Key Differences*
**Feature** | **List** | **Tuple**
**Mutability** | Mutable - you can change, add, remove items | Immutable - cannot change after creation
**Syntax** | `[]` square brackets | `()` round brackets
**Performance** | Slightly slower | Slightly faster, uses less memory
**Methods** | Many: `append`, `extend`, `pop`, `sort` etc | Few: mainly `count`, `index`
**Use case** | When data needs to change | When data should stay constant, like keys for dict
**Can be dict key?** | No | Yes, if elements are also immutable


```python
#### *A. Declaration*
# List
my_list = [1, 2, 3, "hello"]
print(type(my_list)) # <class 'list'>
```

```python
# Tuple
my_tuple = (1, 2, 3, "hello")
print(type(my_tuple)) # <class 'tuple'>
```

```python
# Tuple with 1 item needs a comma
single_tuple = (5,)
#### *B. Mutability*
my_list = [10, 20, 30]
my_list[1] = 99 # Works
my_list.append(40) # Works
print(my_list) # [10, 99, 30, 40]
```

```python
my_tuple = (10, 20, 30)
# my_tuple[1] = 99 # ERROR: 'tuple' object does not support item assignment
# my_tuple.append(40) # ERROR: 'tuple' object has no attribute 'append'
#### *C. Methods*
my_list = [3, 1, 4]
my_list.sort() # [1, 3, 4]
my_list.pop() # removes 4
```

```python
my_tuple = (3, 1, 4, 1)
print(my_tuple.count(1)) # 2
print(my_tuple.index(4)) # 2
```

# No sort, no append, no remove
#### *D. Use as Dictionary Key*
Because tuples are immutable, they can be keys. Lists cannot.
# Works

```python
locations = {
    (6.5244, 3.3792): "Lagos",
    (4.8156, 7.0498): "Port Harcourt"
}
print(locations[(6.5244, 3.3792)]) # Lagos
```

```python
# Doesn't work
# bad_dict = {[1, 2]: "value"} # ERROR: unhashable type: 'list'
```

#### *E. Performance*
import time


```python
# Tuples are faster to iterate
list_data = list(range(1000000))
tuple_data = tuple(range(1000000))

start = time.time()
for i in list_data: pass
print("List:", time.time() - start)

start = time.time()
for i in tuple_data: pass
print("Tuple:", time.time() - start)
```

### *3. When to use which?*

*Use a List when:*
- The collection will change: shopping list, to-do list, scores
students = ["Ada", "Bola"]
students.append("Chike") # add new student
*Use a Tuple when:*
- The data is fixed and shouldn't change: coordinates, RGB color, database record, function returning multiple values
color = (255, 0, 0) # RGB for red - shouldn't change
point = (10.5, 20.3) # x, y coordinate


def get_user():
    return "John", 25, "Port Harcourt" # returns a tuple

name, age, city = get_user()


*Extra tip*: Tuples also protect data. If you pass a tuple to a function, you know it won't be accidentally modified.

### *4. Tuple and List with Unpacking*
Unpacking is where Python really shines with tuples.


```python
# Basic unpacking
point = (10, 20) # tuple
x, y = point
print(x, y) # 10 20
```

```python
# Swapping variables - no temp needed
a, b = 5, 9
a, b = b, a
print(a, b) # 9 5
```

```python
# Works with lists too
data = ["Ada", 25, "PH"]
name, age, city = data
```

```python
# Ignoring values with _
user_info = ("John", 30, "Engineer", 500000)
name, age, *_ = user_info
print(name, age) # John 30
```

*Why tuples are preferred here*: Tuples signal "this structure is fixed". When you see `x, y = point`, you expect 2 items. If it was a list that could change size, it would be risky.

### *5. Nesting: List of Tuples vs Tuple of Lists*
You can nest both, but immutability still applies to the outer container


```python
# List of tuples - very common. Like database rows
students = [
    ("Ada", 90),
    ("Bola", 85),
    ("Chike", 78)
]
students.append(("Dora", 92)) # We can add new student
```

```python
# students[0][1] = 95 # We can also change Bola's score because inner tuple? No!
students[0] = ("Ada", 95) # Must replace the whole tuple
```

```python
# Tuple of lists - outer can't change, inner can
config = (
    ["server1", "server2"],
    ["port80", "port443"]
)
config[0].append("server3") # Works! List inside is mutable
# config[1] = ["new"] # ERROR! Can't replace the tuple item
```

Rule: `tuple` makes the container immutable, not necessarily what's inside it.

### *6. Memory and Performance*
Tuples use less memory and are faster because Python doesn't need to allocate extra space for growth.


```python
import sys

my_list = [1, 2, 3, 4, 5]
my_tuple = (1, 2, 3, 4, 5)

print("List size:", sys.getsizeof(my_list)) # ~120 bytes
print("Tuple size:", sys.getsizeof(my_tuple)) # ~80 bytes
```

Tuples are also hashable if all items are hashable, so they can be used in `set` and as `dict` keys.

```python
# Valid
unique_points = {(1,2), (3,4), (1,2)}
print(unique_points) # {(1, 2), (3, 4)}
```

```python
# Invalid
# bad_set = {[1,2], [3,4]} # ERROR: unhashable type: 'list'
```

### *7. Quick Decision Cheat Sheet*


Situation | Use
Data that changes: add/remove items | **List**
Fixed collection: coordinates, dates | **Tuple**
Dictionary key or set element | **Tuple**
Returning multiple values from function | **Tuple**
Need `.sort()`, `.append()`, `.pop()` | **List**
Want to protect data from accidental change | **Tuple**
### *Summary in 1 line*

Think of it like this:
*List = a notebook you can write in and erase*
*Tuple = a printed paper - read-only*
