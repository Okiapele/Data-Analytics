# 2. Series and DataFrame

Pandas has two main data structures:

## Series
A **Series** is one labelled column of data.

Example: Monthly sales only.

## DataFrame
A **DataFrame** is a full table with rows and columns.

Example: Month, Sales, Returns, Region, Product.

Think of it like Excel:

- One column = Series
- Full spreadsheet = DataFrame

## 2.1 Example 1 — Create a Pandas Series

In this example, we create a single column of monthly sales.

````python
# A Series is a single labelled column of data
sales = pd.Series(
    [4500, 3200, 7800, 5100, 6300],
    index=["Jan", "Feb", "Mar", "Apr", "May"],
    name="Sales"
)

print(sales)
````

### Explanation

`pd.Series(...)` creates a Pandas Series.

The list `[4500, 3200, 7800, 5100, 6300]` contains the actual values.

The `index` labels each value with a month.

The `name="Sales"` gives the Series a name.

This is like one labelled column in Excel.

````python
# Useful Series operations
print("Type:", type(sales))
print("Average sales:", sales.mean())
print("Highest sales:", sales.max())
print("Lowest sales:", sales.min())
print("Total sales:", sales.sum())
print("Month with highest sales:", sales.idxmax())
print("Month with lowest sales:", sales.idxmin())
````

## 2.2 Example 2 — Create Another Series

Now create a Series for student scores.

````python
scores = pd.Series(
    [85, 72, 91, 64, 78],
    index=["Amara", "John", "Priya", "James", "Zara"],
    name="Python Score"
)

print(scores)
print("Average score:", scores.mean())
print("Highest scoring student:", scores.idxmax())
````

# 3. Create a Sample Sales Dataset for This Notebook

The official project uses a file called `sales_dashboard_data.csv`.

To make this notebook runnable even before students receive the official file, the next cell creates a realistic sample sales dataset and saves it as CSV.

When students receive the official dataset, they can replace this file with the official one.

**Note:** The Week 3 project brief lists the 9 core columns, but the required charts include Revenue vs Returns and examples also use Profit. For teaching and practice, this notebook creates `Returns` and `Profit` columns as extra fields so students can practise every slide concept.

````python
import numpy as np
import pandas as pd

np.random.seed(42)

# Create a sample UK retail sales dataset with 1,000 rows
n_rows = 1000

dates = pd.date_range(start="2024-01-01", end="2024-12-31", freq="D")
salespeople = ["Amara", "John", "Priya", "James", "Zara", "Kwame", "Lucy", "Chen"]
regions = ["North", "South", "East", "West"]
products = ["Laptop", "Tablet", "Phone", "Monitor", "Keyboard", "Headset"]

sample_df = pd.DataFrame({
    "OrderID": range(1001, 1001 + n_rows),
    "Date": np.random.choice(dates, size=n_rows),
    "Salesperson": np.random.choice(salespeople, size=n_rows),
    "Region": np.random.choice(regions, size=n_rows),
    "Product": np.random.choice(products, size=n_rows),
    "Qty": np.random.randint(1, 8, size=n_rows),
    "Price": np.random.choice([25, 45, 75, 120, 250, 450, 800], size=n_rows),
    "Discount": np.random.choice([0.00, 0.05, 0.10, 0.15, 0.20], size=n_rows)
})

# Revenue after discount
sample_df["Revenue"] = sample_df["Qty"] * sample_df["Price"] * (1 - sample_df["Discount"])

# Extra useful columns for Week 3 examples
sample_df["Returns"] = np.random.choice([0, 0, 0, 10, 20, 50, 100], size=n_rows)
sample_df["Profit"] = sample_df["Revenue"] * np.random.uniform(0.15, 0.45, size=n_rows)

# Introduce a few messy values intentionally for cleaning demonstration
sample_df.loc[5, "Region"] = " north "
sample_df.loc[10, "Region"] = "SOUTH"
sample_df.loc[15, "Salesperson"] = " priya "
sample_df.loc[20, "Revenue"] = np.nan
sample_df.loc[25, "Product"] = np.nan

# Add a duplicate row intentionally
sample_df = pd.concat([sample_df, sample_df.iloc[[0]]], ignore_index=True)

# Save to CSV
sample_df.to_csv("sales_dashboard_data.csv", index=False)

print("Sample sales_dashboard_data.csv created successfully!")
print("Rows and columns:", sample_df.shape)
sample_df.head()
````

## 5.3 Selecting Rows with `loc`

`loc` is label-based. It uses row labels and column names.

````python
# Row with index label 0
df.loc[0]
````

````python
# Specific cell: row label 0, column Revenue
df.loc[0, "Revenue"]
````

````python
# Rows 0 to 5, selected columns only
df.loc[0:5, ["Date", "Region", "Revenue"]]
````

Important: `loc[0:5]` includes row 5. So it gives rows 0, 1, 2, 3, 4, and 5.

## 5.4 Selecting Rows with `iloc`

`iloc` is position-based. It uses integer positions, like normal Python indexing.

````python
# First row by position
df.iloc[0]
````

````python
# Row 0, column position 3
df.iloc[0, 3]
````

````python
# Rows 0 to 4, columns 1 to 3
df.iloc[0:5, 1:4]
````

Important: `iloc[0:5]` excludes 5. So it gives rows 0, 1, 2, 3, and 4 only.

## 5.5 `loc` vs `iloc` Summary

| Method | Meaning | Uses |
|---|---|---|
| `loc` | label-based | row labels and column names |
| `iloc` | position-based | integer positions |

Example:

```python
df.loc[0, "Revenue"]
df.iloc[0, 8]
```

# 6. Filtering Rows

Filtering means keeping only rows that meet a condition.

Examples:

- Revenue greater than £500
- Region equals North
- Product equals Laptop
- Discount above 10%

````python
# Filter rows where Revenue is greater than 500
high_revenue = df[df["Revenue"] > 500]

high_revenue.head()
````

````python
# Filter rows where Region is North
north_sales = df[df["Region"].str.strip().str.title() == "North"]

north_sales.head()
````

````python
# Filter rows where Product is either Laptop or Phone
selected_products = df[df["Product"].isin(["Laptop", "Phone"])]

selected_products.head()
````

## 6.1 Multiple Conditions

When using multiple conditions:

- Use `&` for AND.
- Use `|` for OR.
- Put each condition inside parentheses.

````python
# Revenue greater than 500 AND Region is North
high_north = df[(df["Revenue"] > 500) & (df["Region"].str.strip().str.title() == "North")]

high_north.head()
````

````python
# Revenue less than 100 OR Returns greater than 50
problem_orders = df[(df["Revenue"] < 100) | (df["Returns"] > 50)]

problem_orders.head()
````

## 6.2 Query Syntax

`.query()` can make some filters easier to read.

````python
# Using query to filter high revenue rows
df.query("Revenue > 500").head()
````

````python
# Query with multiple conditions
df.query("Revenue > 500 and Discount >= 0.10").head()
````

# 8. Adding New Columns — Feature Engineering

Feature engineering means creating new useful columns from existing columns.

This is one of the most valuable analyst skills.

````python
# Create clean text columns first
df["Region"] = df["Region"].str.strip().str.title()
df["Salesperson"] = df["Salesperson"].str.strip().str.title()

# Create calculated columns
df["Gross_Revenue"] = df["Qty"] * df["Price"]
df["Net_Revenue"] = df["Gross_Revenue"] * (1 - df["Discount"])
df["Profit_Margin"] = (df["Profit"] / df["Revenue"]) * 100

df[["Qty", "Price", "Discount", "Gross_Revenue", "Net_Revenue", "Profit_Margin"]].head()
````

## 8.1 Revenue Bands with `pd.cut()`

`pd.cut()` is used to convert numbers into categories.

````python
df["Revenue_Band"] = pd.cut(
    df["Revenue"],
    bins=[0, 200, 500, 1000, 5000],
    labels=["Low", "Medium", "High", "Very High"]
)

df[["Revenue", "Revenue_Band"]].head(10)
````

# 9. Date Handling with `.dt`

Dates are very important in data analytics.

We often need to extract year, month, day, weekday, and quarter.

````python
# Make sure Date is datetime
df["Date"] = pd.to_datetime(df["Date"])

# Extract date parts
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.strftime("%b")
df["Day"] = df["Date"].dt.day
df["Weekday"] = df["Date"].dt.day_name()
df["Quarter"] = df["Date"].dt.quarter
df["Is_Weekend"] = df["Date"].dt.dayofweek >= 5

df[["Date", "Year", "Month", "Month_Name", "Day", "Weekday", "Quarter", "Is_Weekend"]].head()
````

````python
# Filter Quarter 1 sales
q1_sales = df[df["Quarter"] == 1]
q1_sales.head()
````

````python
# Filter January sales
jan_sales = df[df["Month"] == 1]
jan_sales.head()
````

# 10. Renaming, Dropping, and Changing Data Types

In real projects, you often need to rename columns, remove temporary columns, and convert data types.

````python
# Example: create a copy so we do not damage the main df
df_types = df.copy()

# Rename a column
df_types.rename(columns={"Revenue": "Revenue_GBP"}, inplace=True)

# Convert OrderID to string because we do not calculate with IDs
df_types["OrderID"] = df_types["OrderID"].astype(str)

# Convert Region to category for memory efficiency
df_types["Region"] = df_types["Region"].astype("category")

df_types.info()
````

# 11. GroupBy and Aggregation

GroupBy is like a Pivot Table in Excel.

It follows this pattern:

**Split → Apply → Combine**

Example: For each Region, calculate total Revenue.

````python
# Total revenue by region
region_revenue = df.groupby("Region")["Revenue"].sum()
region_revenue
````

````python
# Average revenue by product
product_average = df.groupby("Product")["Revenue"].mean()
product_average
````

````python
# Count orders by salesperson
orders_by_salesperson = df.groupby("Salesperson")["OrderID"].count()
orders_by_salesperson
````

## 11.1 Multiple Aggregations with `.agg()`

`.agg()` allows us to calculate many summary statistics at once.

````python
region_summary = df.groupby("Region").agg({
    "Revenue": ["sum", "mean", "count", "max"],
    "Returns": "sum",
    "Profit": "mean"
})

region_summary
````

## 11.2 Reset Index

After GroupBy, the grouped column may become the index.

Use `reset_index()` to turn it back into a normal column.

````python
region_summary_flat = df.groupby("Region").agg(
    Total_Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean"),
    Number_of_Orders=("OrderID", "count"),
    Total_Returns=("Returns", "sum"),
    Average_Profit=("Profit", "mean")
).reset_index()

region_summary_flat
````

## 11.3 Group By Multiple Columns

Example: What is the total revenue by Month and Region?

````python
monthly_region = df.groupby(["Month", "Region"])["Revenue"].sum().reset_index()
monthly_region.head(12)
````

## 11.4 Transform

`transform()` adds group-level statistics back to every original row.

````python
df["Region_Avg_Revenue"] = df.groupby("Region")["Revenue"].transform("mean")
df["Pct_of_Region_Avg"] = (df["Revenue"] / df["Region_Avg_Revenue"]) * 100

df[["Region", "Revenue", "Region_Avg_Revenue", "Pct_of_Region_Avg"]].head()
````

# 12. Merging DataFrames

Merging means combining two tables using a common column.

This is like VLOOKUP in Excel.

````python
orders = pd.DataFrame({
    "OrderID": [101, 102, 103, 104, 105],
    "CustomerID": [1, 2, 1, 3, 5],
    "Sales": [450, 320, 780, 210, 999]
})

customers = pd.DataFrame({
    "CustomerID": [1, 2, 3, 4],
    "Name": ["Alice", "Bob", "Carol", "David"],
    "Region": ["North", "South", "East", "West"]
})

print("Orders table:")
display(orders)

print("Customers table:")
display(customers)
````

````python
# Inner join: only matching CustomerID values in both tables
inner_merged = pd.merge(orders, customers, on="CustomerID", how="inner")
inner_merged
````

````python
# Left join: keep all orders, add customer details where found
left_merged = pd.merge(orders, customers, on="CustomerID", how="left")
left_merged
````

````python
# Outer join: keep all rows from both tables
outer_merged = pd.merge(orders, customers, on="CustomerID", how="outer")
outer_merged
````

## 12.1 Join Types Summary

| Join Type | Meaning |
|---|---|
| Inner | Keep only matching rows in both tables |
| Left | Keep all rows from left table |
| Right | Keep all rows from right table |
| Outer | Keep all rows from both tables |

If there is no match, Pandas fills missing values with `NaN`.

# 13. Data Cleaning with Pandas

Real-world data is usually messy.

Common problems:

- Missing values
- Duplicate rows
- Wrong data types
- Extra spaces
- Inconsistent capitalisation
- Numbers stored as text
- Dates stored as text

Data cleaning makes messy data usable.

## 13.1 Find the Problems

````python
print("Missing values per column:")
print(df.isnull().sum())

print("
Duplicate rows:")
print(df.duplicated().sum())

print("
Data types:")
print(df.dtypes)

print("
Region value counts:")
print(df["Region"].value_counts())
````

## 13.2 Handle Missing Values

There is no single correct answer for missing values. The right decision depends on the context.

Common strategies:

- Drop rows
- Drop columns
- Fill with 0
- Fill with Unknown
- Fill with median or mean
- Forward fill for time-series

````python
# Create a copy for cleaning examples
df_clean = df.copy()

print("Before cleaning:")
print(df_clean.isnull().sum())
````

````python
# Fill missing Product values with Unknown
df_clean["Product"] = df_clean["Product"].fillna("Unknown")

# Fill missing Revenue values with the median
df_clean["Revenue"] = df_clean["Revenue"].fillna(df_clean["Revenue"].median())

print("After filling missing values:")
print(df_clean.isnull().sum())
````

## 13.3 Remove Duplicates

````python
print("Duplicates before:", df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print("Duplicates after:", df_clean.duplicated().sum())
````

## 13.4 Fix Text Formatting

````python
df_clean["Region"] = df_clean["Region"].str.strip().str.title()
df_clean["Salesperson"] = df_clean["Salesperson"].str.strip().str.title()
df_clean["Product"] = df_clean["Product"].str.strip().str.title()

df_clean[["Region", "Salesperson", "Product"]].head(10)
````

## 13.5 Change Data Types

````python
df_clean["OrderID"] = df_clean["OrderID"].astype(str)
df_clean["Date"] = pd.to_datetime(df_clean["Date"])
df_clean["Revenue"] = df_clean["Revenue"].astype(float)

df_clean.info()
````

## 13.6 Cleaning Decision Guide

Use this simple guide:

- Less than 5% missing numeric values: consider filling with mean or median.
- 5–20% missing: investigate and consider filling with Unknown or median.
- More than 20% missing: consider whether the column is reliable.
- Duplicate rows: investigate and remove when appropriate.
- Numbers stored as text: convert using `.astype(float)` or `pd.to_numeric()`.
- Dates stored as text: convert using `pd.to_datetime()`.
- Text with inconsistent spacing/case: use `.str.strip()`, `.str.title()`, `.str.lower()`.

## 14.1 Line Chart — Monthly Revenue Trend

Use a line chart when you want to show change over time.

````python
monthly_revenue = df_clean.groupby("Month")["Revenue"].sum().reset_index()

month_labels = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

plt.figure(figsize=(10, 5))
plt.plot(monthly_revenue["Month"], monthly_revenue["Revenue"], marker="o", linewidth=2)
plt.xticks(range(1, 13), month_labels)
plt.xlabel("Month")
plt.ylabel("Revenue (£)")
plt.title("Monthly Revenue Trend 2024")
plt.tight_layout()
plt.show()
````

### Interpretation Example

The line chart shows how total revenue changes month by month. A rising pattern may suggest growth, seasonal demand, or effective campaigns. A falling pattern may suggest weaker demand, stock issues, or reduced customer activity.

## 15.1 Correlation Heatmap

A correlation heatmap shows relationships between numeric variables.

Correlation values range from:

- `1`: strong positive relationship
- `0`: no linear relationship
- `-1`: strong negative relationship

````python
corr = df_clean[["Qty", "Price", "Discount", "Revenue", "Returns", "Profit"]].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
````

## 15.2 Box Plot — Revenue by Product

A box plot shows median, spread, outliers, and distribution by category.

````python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_clean, x="Product", y="Revenue")
plt.title("Revenue Distribution by Product")
plt.xlabel("Product")
plt.ylabel("Revenue per Order (£)")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
````

## 15.3 Seaborn Barplot — Average Revenue by Product

A Seaborn barplot can calculate and show average values automatically.

````python
plt.figure(figsize=(10, 5))
sns.barplot(data=df_clean, x="Product", y="Revenue", estimator="mean")
plt.title("Average Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Average Revenue (£)")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
````

# 16. Subplots — Building a Dashboard

A dashboard is a collection of charts in one figure.

Use:

```python
fig, axes = plt.subplots(2, 2)
```

This creates a 2 by 2 grid of charts.

````python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("Sales Performance Dashboard 2024", fontsize=16, fontweight="bold")

# Chart 1: Monthly revenue trend
monthly = df_clean.groupby("Month")["Revenue"].sum()
axes[0, 0].plot(monthly.index, monthly.values, marker="o", linewidth=2)
axes[0, 0].set_title("Monthly Revenue Trend")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Revenue (£)")

# Chart 2: Revenue by region
region_s = df_clean.groupby("Region")["Revenue"].sum().sort_values(ascending=False)
axes[0, 1].bar(region_s.index, region_s.values)
axes[0, 1].set_title("Revenue by Region")
axes[0, 1].set_xlabel("Region")
axes[0, 1].set_ylabel("Revenue (£)")

# Chart 3: Returns distribution
axes[1, 0].hist(df_clean["Returns"], bins=20, edgecolor="black")
axes[1, 0].set_title("Returns Distribution")
axes[1, 0].set_xlabel("Returns (£)")
axes[1, 0].set_ylabel("Count")

# Chart 4: Revenue vs Returns
axes[1, 1].scatter(df_clean["Returns"], df_clean["Revenue"], alpha=0.5)
axes[1, 1].set_title("Revenue vs Returns")
axes[1, 1].set_xlabel("Returns (£)")
axes[1, 1].set_ylabel("Revenue (£)")

plt.tight_layout()
plt.savefig("week3_dashboard_example.png", dpi=150, bbox_inches="tight")
plt.show()
````

# 17. Chart Design Principles

A chart must communicate clearly. A technically correct chart is not enough.

Professional chart principles:

1. Title tells the story.
2. Label everything.
3. Choose the right chart.
4. Use colour with purpose.
5. Remove chartjunk.
6. Write an interpretation.

For Project 2, every chart must have a written interpretation underneath.

## 17.1 Good vs Weak Chart Titles

Weak title:

```text
Sales Chart
```

Better title:

```text
North Region Generated the Highest Revenue in 2024
```

The better title already communicates the insight.

## 17.2 Chart Selection Guide

| Business Question | Best Chart |
|---|---|
| How did revenue change over time? | Line chart |
| Which region performed best? | Bar chart |
| Are revenue and returns related? | Scatter plot |
| What is the most common order value? | Histogram |
| Which months and regions performed best? | Heatmap |
| Which product has the most outliers? | Box plot |

# 19. Project 2 Full Starter Code

The following section gives a full structured version of the dashboard project.

## 19.1 Cell 1 — Import Libraries

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
````

## 19.2 Cell 2 — Load the Dataset

````python
df_project = pd.read_csv("sales_dashboard_data.csv", parse_dates=["Date"])

print("Dataset loaded successfully!")
print("Shape:", df_project.shape)
df_project.head()
````

## 19.3 Cell 3 — Inspect the Dataset

````python
print("First 5 rows:")
display(df_project.head())

print("
Shape:")
print(df_project.shape)

print("
Info:")
df_project.info()

print("
Missing values:")
print(df_project.isnull().sum())

print("
Statistical summary:")
display(df_project.describe())
````

## 19.4 Cell 4 — Clean the Data

````python
df_project = df_project.copy()

# Remove duplicates
df_project.drop_duplicates(inplace=True)

# Fill missing values
df_project["Product"] = df_project["Product"].fillna("Unknown")
df_project["Revenue"] = df_project["Revenue"].fillna(df_project["Revenue"].median())

# Clean text columns
df_project["Region"] = df_project["Region"].str.strip().str.title()
df_project["Salesperson"] = df_project["Salesperson"].str.strip().str.title()
df_project["Product"] = df_project["Product"].str.strip().str.title()

# Convert data types
df_project["Date"] = pd.to_datetime(df_project["Date"])
df_project["OrderID"] = df_project["OrderID"].astype(str)

print("Clean dataset shape:", df_project.shape)
print("Missing values after cleaning:")
print(df_project.isnull().sum())
````

## 19.5 Cell 5 — Feature Engineering

````python
df_project["Month"] = df_project["Date"].dt.month
df_project["Month_Name"] = df_project["Date"].dt.strftime("%b")
df_project["Quarter"] = df_project["Date"].dt.quarter
df_project["Gross_Revenue"] = df_project["Qty"] * df_project["Price"]
df_project["Net_Revenue"] = df_project["Gross_Revenue"] * (1 - df_project["Discount"])
df_project["Revenue_Band"] = pd.cut(
    df_project["Revenue"],
    bins=[0, 200, 500, 1000, 5000],
    labels=["Low", "Medium", "High", "Very High"]
)

df_project.head()
````

# 20. Project Chart 1 — Monthly Revenue Trend

Business question: How did revenue change month by month?

Chart type: Line chart, because we are showing change over time.

````python
monthly = df_project.groupby("Month")["Revenue"].sum().reset_index()

month_labels = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(
    monthly["Month"],
    monthly["Revenue"],
    marker="o",
    linewidth=2.5,
    markersize=6
)

ax.fill_between(monthly["Month"], monthly["Revenue"], alpha=0.1)

ax.set_xticks(range(1, 13))
ax.set_xticklabels(month_labels)
ax.set_title("Monthly Revenue Trend 2024", fontsize=14, fontweight="bold")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue (£)")

plt.tight_layout()
plt.savefig("chart1_monthly_revenue_trend.png", dpi=150, bbox_inches="tight")
plt.show()
````

## Chart 1 Interpretation Example

The line chart shows how revenue changed across the 12 months of 2024. The highest month represents the strongest sales period, while the lowest month may indicate weaker demand or fewer orders. This insight can help the business plan stock, staffing, and marketing campaigns around strong and weak months.

# 21. Project Chart 2 — Revenue by Region

Business question: Which region generated the most revenue?

Chart type: Bar chart, because we are comparing categories.

````python
region_rev = df_project.groupby("Region")["Revenue"].sum().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(region_rev.index, region_rev.values)

ax.bar_label(
    bars,
    labels=[f"£{v:,.0f}" for v in region_rev.values],
    padding=3
)

ax.set_title("Total Revenue by Region", fontsize=13, fontweight="bold")
ax.set_xlabel("Region")
ax.set_ylabel("Revenue (£)")

plt.tight_layout()
plt.savefig("chart2_revenue_by_region.png", dpi=150, bbox_inches="tight")
plt.show()
````

## Chart 2 Interpretation Example

The bar chart compares total revenue across all regions. The region with the tallest bar generated the highest revenue, while the shortest bar indicates the weakest regional performance. This can help management identify where sales strategies are working and where additional support may be needed.

# 22. Project Chart 3 — Revenue vs Returns

Business question: Is there any visible relationship between revenue and returns?

Chart type: Scatter plot, because we are comparing two numeric variables.

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(df_project["Returns"], df_project["Revenue"], alpha=0.6)

ax.set_title("Revenue vs Returns", fontsize=13, fontweight="bold")
ax.set_xlabel("Returns (£)")
ax.set_ylabel("Revenue (£)")

plt.tight_layout()
plt.savefig("chart3_revenue_vs_returns.png", dpi=150, bbox_inches="tight")
plt.show()
````

## Add a Trend Line

A trend line can help us see the general direction of the relationship.

````python
# Remove missing values for trend line calculation
scatter_df = df_project[["Returns", "Revenue"]].dropna()

x = scatter_df["Returns"]
y = scatter_df["Revenue"]

# Calculate trend line using numpy
coefficients = np.polyfit(x, y, 1)
trend_line = np.poly1d(coefficients)

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(x, y, alpha=0.5, label="Orders")
ax.plot(x, trend_line(x), linewidth=2, label="Trend line")

ax.set_title("Revenue vs Returns with Trend Line", fontsize=13, fontweight="bold")
ax.set_xlabel("Returns (£)")
ax.set_ylabel("Revenue (£)")
ax.legend()

plt.tight_layout()
plt.savefig("chart3_revenue_vs_returns_trendline.png", dpi=150, bbox_inches="tight")
plt.show()
````

## Chart 3 Interpretation Example

The scatter plot shows the relationship between revenue and returns. If the trend line slopes upward, it may suggest that higher revenue orders are associated with higher returns. If there is no clear pattern, returns may not be strongly linked to revenue and may need further investigation by product or region.

# 23. Project Chart 4 — Monthly Performance Heatmap

Business question: Which months and regions generated the strongest revenue?

Chart type: Heatmap, because we want to compare two categories using colour intensity.

````python
pivot = df_project.pivot_table(
    values="Revenue",
    index="Month",
    columns="Region",
    aggfunc="sum"
)

fig, ax = plt.subplots(figsize=(10, 7))

sns.heatmap(
    pivot,
    annot=True,
    fmt=",.0f",
    cmap="YlOrRd",
    linewidths=0.5,
    cbar_kws={"label": "Revenue (£)"},
    ax=ax
)

ax.set_title("Monthly Revenue by Region Heatmap", fontsize=13, fontweight="bold")
ax.set_xlabel("Region")
ax.set_ylabel("Month")

plt.tight_layout()
plt.savefig("chart4_monthly_region_heatmap.png", dpi=150, bbox_inches="tight")
plt.show()
````

## Chart 4 Interpretation Example

The heatmap shows revenue performance by month and region. Darker colours indicate stronger revenue performance, while lighter colours show weaker performance. This chart helps the business quickly identify which region-month combinations performed best and where seasonal or regional weaknesses exist.

# 24. Project Chart 5 — Order Value Distribution

Business question: What is the most common order value range?

Chart type: Histogram, because we are showing the distribution of a numeric variable.

````python
fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(df_project["Revenue"], bins=25, edgecolor="black")

ax.set_title("Order Value Distribution", fontsize=13, fontweight="bold")
ax.set_xlabel("Revenue per Order (£)")
ax.set_ylabel("Number of Orders")

plt.tight_layout()
plt.savefig("chart5_order_value_distribution.png", dpi=150, bbox_inches="tight")
plt.show()
````

## Chart 5 Interpretation Example

The histogram shows how order values are distributed. If most bars are concentrated on the lower end, it means most customers place smaller orders. If there are a few bars far to the right, those may represent high-value orders that deserve closer business attention.

# 25. Project Chart 6 — Revenue by Product

Business question: Which products have the highest revenue spread and the most outliers?

Chart type: Box plot, because it shows spread, median, and outliers.

````python
fig, ax = plt.subplots(figsize=(12, 6))

sns.boxplot(
    data=df_project,
    x="Product",
    y="Revenue",
    ax=ax
)

ax.set_title("Revenue Distribution by Product", fontsize=13, fontweight="bold")
ax.set_xlabel("Product Category")
ax.set_ylabel("Revenue per Order (£)")
plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.savefig("chart6_revenue_by_product_boxplot.png", dpi=150, bbox_inches="tight")
plt.show()
````

## Chart 6 Interpretation Example

The box plot compares revenue distribution across product categories. Products with wider boxes or longer whiskers show greater variation in order value. Outliers may represent unusually high-value orders and should be investigated for sales opportunities or unusual behaviour.
